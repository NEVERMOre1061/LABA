```python
#Задание №2
import os

# 1 и 2. Создайте и запишите текстовый файл (используем with open, который автоматически закроет файл после записи)
with open("example.txt", "w", encoding="utf-8") as f:
    f.write("Первая строка\n")
    f.write("Вторая строка\n")

print("Файл создан и в него записаны строки.")

# 3. Откройте файл снова, считайте содержимое и выведите на экран
with open("example.txt", "r", encoding="utf-8") as f:
    content = f.read()
    print("Содержимое файла:\n", content)

# 4. Проверьте, существует ли файл, используя os.path.exists()
file_exists = os.path.exists("example.txt")
print("Файл существует:", file_exists)

# 5. Удалите файл с помощью модуля os
if file_exists:
    os.remove("example.txt")
    print("Файл удален.")

# Проверка удаления
print("Файл существует после удаления:", os.path.exists("example.txt"))
```

```python
# 1. Подключение Google Drive к Colab
from google.colab import drive
drive.mount('/content/drive')

# Проверка подключения диска
print("\n--- Проверка папки на Диске ---")
!ls /content/drive/MyDrive

# 2. Создание и запись текста в файл на Google Drive
file_path = "/content/drive/MyDrive/my_text_file.txt"
with open(file_path, "w", encoding="utf-8") as f:
    f.write("Это частный файл, сохраненный в Google Drive.\n")
    f.write("Вторая строка текста.\n")

print("\nФайл успешно сохранен в Google Drive.")

# 3. Чтение файла из Google Drive
with open(file_path, "r", encoding="utf-8") as f:
    content = f.read()
    print("Содержимое файла из Google Drive:\n", content)

# 4. Создание CSV-файла со списком студентов вручную
students = [
    ["Иванов И.И.", "20", "ИВТ-101"],
    ["Петров П.П.", "22", "ИВТ-102"],
    ["Сидорова А.А.", "21", "ИВТ-103"]
]

csv_path = "/content/drive/MyDrive/students.csv"
with open(csv_path, "w", encoding="utf-8") as f:
    for student in students:
        f.write(",".join(student) + "\n")

print("\nФайл students.csv успешно сохранен в Google Drive.")
```
