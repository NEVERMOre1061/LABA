```python
## ИДЗ 6
from datetime import datetime
import pandas as pd

# Исходный DataFrame с примерными данными
trains_df = pd.DataFrame([
    {'Пункт назначения': 'Москва', 'Номер поезда': '002А', 'Время отправления': '08:30'},
    {'Пункт назначения': 'Санкт-Петербург', 'Номер поезда': '016А', 'Время отправления': '12:15'},
    {'Пункт назначения': 'Казань', 'Номер поезда': '050Г', 'Время отправления': '18:45'},
    {'Пункт назначения': 'Екатеринбург', 'Номер поезда': '012Е', 'Время отправления': '21:10'},
])

# Ввод новых записей с клавиатуры (если нужно добавить строки)
add_more = input("Добавить новые записи о поездах? (да/нет): ").strip().lower()
while add_more in ('да', 'y', 'yes', 'д'):
    dest = input("Введите пункт назначения: ").strip()
    train_num = input("Введите номер поезда: ").strip()
    dep_time = input("Введите время отправления (ЧЧ:ММ): ").strip()

    # Простая проверка формата времени ЧЧ:ММ
    try:
        datetime.strptime(dep_time, "%H:%M")
        new_row = pd.DataFrame([{
            'Пункт назначения': dest,
            'Номер поезда': train_num,
            'Время отправления': dep_time
        }])
        trains_df = pd.concat([trains_df, new_row], ignore_index=True)
    except ValueError:
        print("Неверный формат времени. Используйте ЧЧ:ММ (например, 14:20).")

    add_more = input("Добавить еще одну запись? (да/нет): ").strip().lower()

# Упорядочивание записей в алфавитном порядке по названиям пунктов назначения
trains_sorted = trains_df.sort_values(by='Пункт назначения', ascending=True).reset_index(drop=True)
print("\nВсе записи, отсортированные по пункту назначения (А–Я):")
print(trains_sorted.to_string(index=False))

# Ввод порогового времени и фильтрация поездов, отправляющихся после него
filter_time_str = input("\nВведите время отправления для фильтрации (ЧЧ:ММ): ").strip()
try:
    filter_time = datetime.strptime(filter_time_str, "%H:%M").time()

    mask = trains_sorted['Время отправления'].apply(
        lambda t: datetime.strptime(t, "%H:%M").time() > filter_time
    )
    filtered_trains = trains_sorted[mask]

    print(f"\nПоезда, отправляющиеся после {filter_time_str}:")
    if filtered_trains.empty:
        print("Поездов, отправляющихся после указанного времени, нет.")
    else:
        print(filtered_trains.to_string(index=False))
except ValueError:
    print("Ошибка: введите время в формате ЧЧ:ММ (например, 09:00).")





```
