```python
## Задание №1
import numpy as np

# 1. Создание массива 3x3 от 1 до 9
arr1 = np.arange(1, 10).reshape(3, 3)
print("Исходный массив 3x3:\n", arr1)

# 2. Умножение всех элементов на 2
arr1 = arr1 * 2
print("\nМассив после умножения на 2:\n", arr1)

# 3. Замена всех элементов больше 10 на 0
arr1[arr1 > 10] = 0

# 4. Вывод итогового массива
print("\nИтоговый массив:\n", arr1)
```

```python
## Задание № 2
import numpy as np

# 1. Создание массива из 20 случайных целых чисел от 1 до 100
arr2 = np.random.randint(1, 101, size=20)
print("Исходный массив:", arr2)

# 2. Поиск и вывод элементов, делящихся на 5 без остатка
divisible_by_5 = arr2[arr2 % 5 == 0]
print("\nЭлементы, делящиеся на 5:", divisible_by_5)

# 3. Замена элементов, делящихся на 5, на -1
arr2[arr2 % 5 == 0] = -1

# 4. Вывод обновленного массива
print("\nОбновленный массив:", arr2)
```

```python
## Задание № 3

import numpy as np

# 1. Создание двух массивов 1x5 со случайными числами от 0 до 50
a = np.random.randint(0, 51, size=5)
b = np.random.randint(0, 51, size=5)
print("Массив A:", a)
print("Массив B:", b)

# 2. Объединение в один двумерный массив (по строкам)
combined = np.vstack((a, b))
print("\nОбъединенный двумерный массив:\n", combined)

# 3. Разделение полученного массива на два обратно
part1, part2 = np.vsplit(combined, 2)
print("\nРазделенная часть 1:\n", part1)
print("Разделенная часть 2:\n", part2)
```

```python
## Задание № 4

import numpy as np

# 1. Массив из 50 чисел, равномерно распределенных от -10 до 10
arr4 = np.linspace(-10, 10, 50)
print("Сгенерированный массив из 50 чисел:\n", arr4)

# 2. Вычисления
total_sum = np.sum(arr4)
pos_sum = np.sum(arr4[arr4 > 0])
neg_sum = np.sum(arr4[arr4 < 0])

# 3. Вывод результатов
print(f"\nСумма всех элементов: {total_sum:.4f}")
print(f"Сумма положительных элементов: {pos_sum:.4f}")
print(f"Сумма отрицательных элементов: {neg_sum:.4f}")
```

```python
## Задание № 5

import numpy as np

# 1. Единичная матрица 4x4
identity_matrix = np.eye(4)
print("Единичная матрица 4x4:\n", identity_matrix)

# 2. Диагональная матрица 4x4 с элементами [5, 10, 15, 20] (без циклов)
diag_matrix = np.diag([5, 10, 15, 20])
print("\nДиагональная матрица 4x4:\n", diag_matrix)

# 3. Сумма элементов каждой матрицы и сравнение
sum_identity = np.sum(identity_matrix)
sum_diag = np.sum(diag_matrix)

print(f"\nСумма элементов единичной матрицы: {sum_identity}")
print(f"Сумма элементов диагональной матрицы: {sum_diag}")

if sum_identity > sum_diag:
    print("Сумма единичной матрицы больше.")
elif sum_diag > sum_identity:
    print("Сумма диагональной матрицы больше.")
else:
    print("Суммы равны.")
```

```python
## Задание № 8

import numpy as np

# Создание случайной квадратной матрицы 3x3
square_matrix = np.random.randint(1, 10, size=(3, 3))
print("Исходная матрица 3x3:\n", square_matrix)

# 1. Определитель этой матрицы
det = np.linalg.det(square_matrix)
print(f"\nОпределитель матрицы: {det:.4f}")

# 2. Обратная матрица (или проверка на вырожденность)
try:
    inv_matrix = np.linalg.inv(square_matrix)
    print("\nОбратная матрица:\n", inv_matrix)
except np.linalg.LinAlgError:
    print("\nМатрица вырождена, обратной матрицы не существует.")
```

```python
## Задание № 10

import numpy as np

# Матрица коэффициентов при неизвестных (x, y, z)
A = np.array([
    [2, 3, -1],
    [4, -1, 2],
    [-3, 5, 4]
])

# Вектор правой части (вободные члены)
B = np.array([5, 6, -2])

print("Матрица коэффициентов A:\n", A)
print("\nВектор свободных членов B:", B)

# Решение системы уравнений с помощью np.linalg.solve
solution = np.linalg.solve(A, B)

print("\nРезультат решения (значения x, y, z):")
print(f"x = {solution[0]:.4f}")
print(f"y = {solution[1]:.4f}")
print(f"z = {solution[2]:.4f}")


```

```python
## Индивидуальное Задание № 11

#Дано x- Кол-во воды в 1 резервуаре, y - кол-во воды во 2 резервуаре, z- кол-во воды в резервуаре 3.
# 1) В первый резервуар поступает в два раза больше воды чем во 2 x=2y ->> x-2y=0
# 2) В третий на 50 литров больше чем во второй z=y+50 ->> -y+z =50
# 1) Всего в систему поступает 500 литров воды x+y+z=500
# В матричном виде A*X=B
# Матрица A:
# [1 -2 0]
# [0 -1 1]
# [1 1  1]

import numpy as np

# 1. Задаем матрицу коэффициентов (A) и вектор свободных членов (B)
A = np.array([
    [1, -2, 0],
    [0, -1, 1],
    [1, 1, 1]
], dtype=float)

B = np.array([0, 50, 500], dtype=float)

print("Матрица коэффициентов A:\n", A)
print("Вектор B:", B)

# --- Способ 1: С помощью np.linalg.solve (по заданию) ---
solution_solve = np.linalg.solve(A, B)
print("\n--- Результат через np.linalg.solve ---")
print(f"Решение: x = {solution_solve[0]:.2f}, y = {solution_solve[1]:.2f}, z = {solution_solve[2]:.2f}")

# --- Способ 2: Матричный метод (X = A^(-1) * B) ---
inv_A = np.linalg.inv(A)
solution_matrix = inv_A.dot(B)
print("\n--- Результат матричным методом ---")
print(f"Решение: x = {solution_matrix[0]:.2f}, y = {solution_matrix[1]:.2f}, z = {solution_matrix[2]:.2f}")

# --- Способ 3: Метод Крамера ---
det_A = np.linalg.det(A)
if det_A != 0:
    # Создаем вспомогательные матрицы, заменяя столбцы матрицы A на вектор B
    A1 = A.copy()
    A1[:, 0] = B

    A2 = A.copy()
    A2[:, 1] = B

    A3 = A.copy()
    A3[:, 2] = B

    x_kramer = np.linalg.det(A1) / det_A
    y_kramer = np.linalg.det(A2) / det_A
    z_kramer = np.linalg.det(A3) / det_A

    print("\n--- Результат методом Крамера ---")
    print(f"Решение: x = {x_kramer:.2f}, y = {x_kramer:.2f}, z = {z_kramer:.2f}")
else:
    print("\nОпределитель матрицы равен нулю, методом Крамера решить нельзя.")

print(f"В первом резервуаре: {solution_solve[0]:.2f} литров")
print(f"Во втором резервуаре: {solution_solve[1]:.2f} литров")
print(f"В третьем резервуаре: {solution_solve[2]:.2f} литров")



```
