```python
## Задание 1 Создание Series из списка
import pandas as pd

s1 = pd.Series([5, 15, 25, 35, 45], index=['a', 'b', 'c', 'd', 'e'])

print("Series:\n", s1)
print("\nТип данных:", s1.dtype)
```

```python
## Задание 2 получение элемента Series
import pandas as pd

s2 = pd.Series([12, 24, 36, 48, 60], index=['A', 'B', 'C', 'D', 'E'])

print("Элемент с индексом 'C' (.loc):", s2.loc['C'])
print("Третий элемент (.iloc):", s2.iloc[2])
```

```python
## Задание 3 фильтрация данных с помощью логической индексации
import numpy as np
import pandas as pd

s3 = pd.Series(np.array([4, 9, 16, 25, 36, 49, 64]))
filtered_s3 = s3[s3 > 20]

print("Элементы больше 20:\n", filtered_s3)
```

```python
## Задание 4 просмотр первых и последних элементов
import numpy as np
import pandas as pd

s4 = pd.Series(np.random.randint(1, 101, size=50))

print("Первые 7 элементов:\n", s4.head(7))
print("\nПоследние 5 элементов:\n", s4.tail(5))
```

```python
## Задание 5 определение типа данных Series
import pandas as pd

s5 = pd.Series(['cat', 'dog', 'rabbit', 'parrot', 'fish'])

print("Исходный тип данных:", s5.dtype)

s5_cat = s5.astype('category')
print("\nПосле преобразования в category:\n", s5_cat)
print("Новый тип данных:", s5_cat.dtype)
```

```python
## Задание 6 проверка пропущенных значений
import numpy as np
import pandas as pd

s6 = pd.Series([1.2, np.nan, 3.4, np.nan, 5.6, 6.8])

has_nan = s6.isna().any()
nan_indices = s6.index[s6.isna()].tolist()

print("Есть ли пропущенные значения:", has_nan)
print("Индексы пропущенных элементов:", nan_indices)
```

```python
## Задание 7 заполнение пропцущенных значений
import numpy as np
import pandas as pd

s7 = pd.Series([1.2, np.nan, 3.4, np.nan, 5.6, 6.8])

mean_val = s7.mean()
s7_filled = s7.fillna(mean_val)

print(f"Среднее значение непустых элементов: {mean_val:.4f}")
print("\nSeries после заполнения NaN:\n", s7_filled)
```

```python
## Задание 8 арифметические операции с Series
import pandas as pd

s1 = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
s2 = pd.Series([5, 15, 25, 35], index=['b', 'c', 'd', 'e'])

res = s1 + s2
print("Результат сложения с NaN:\n", res)

# Почему появляются NaN:
# pandas сопоставляет элементы по индексам. Индекса 'a' нет в s2,
# а индекса 'e' нет в s1 — для несовпадающих индексов ставится NaN.

res_filled = res.fillna(0)
print("\nРезультат после замены NaN на 0:\n", res_filled)
```

```python
## Задание 9 Применение функции Series
import numpy as np
import pandas as pd

s9 = pd.Series([2, 4, 6, 8, 10])
s9_sqrt = s9.apply(np.sqrt)

print("Исходный Series:\n", s9)
print("\nКвадратный корень элементов:\n", s9_sqrt)
```

```python
## Задание 10 основные статистические методы
import numpy as np
import pandas as pd

s10 = pd.Series(np.random.randint(50, 151, size=20))

print("Series:\n", s10)
print("\nСумма:", s10.sum())
print("Среднее:", s10.mean())
print("Минимум:", s10.min())
print("Максимум:", s10.max())
print("Стандартное отклонение:", s10.std())
```

```python
## Задание 12  проверка уникальных индексов
import pandas as pd

s12 = pd.Series([10, 20, 30, 40, 50, 60], index=['A', 'B', 'A', 'C', 'D', 'B'])

is_unique = s12.index.is_unique
print("Индексы уникальны:", is_unique)

if not is_unique:
    grouped = s12.groupby(level=0).sum()
    print("\nСгруппированные и сложенные по индексу значения:\n", grouped)
```

```python
## Задание 15 построение графика на основе Series
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Даты с 1 по 30 марта 2024 года (30 дней)
dates_march = pd.date_range(start="2024-03-01", periods=30, freq="D")
values_march = np.random.randint(50, 151, size=30)

s15 = pd.Series(values_march, index=dates_march)

plt.figure(figsize=(10, 5))
plt.plot(s15.index, s15.values, marker="o")

plt.title("Значения за март 2024 года")
plt.xlabel("Дата")
plt.ylabel("Значение")
plt.grid(True)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
## Индивидуальное задание вариант 6
import pandas as pd
import matplotlib.pyplot as plt

# 1. Создаем CSV-файл sales.csv
sales_csv = """Дата,Продажи
2024-07-01,120
2024-07-02,150
2024-07-03,170
2024-07-04,160
2024-07-05,180"""

with open("sales.csv", "w", encoding="utf-8") as f:
    f.write(sales_csv)

# 2. Читаем данные и устанавливаем DatetimeIndex
df = pd.read_csv("sales.csv")
df["Дата"] = pd.to_datetime(df["Дата"])
df = df.set_index("Дата")
s_sales = df["Продажи"]

# 3. Находим скользящее среднее
s_rolling = s_sales.rolling(2).mean()

print("Реальные продажи:\n", s_sales)
print("\nСкользящее среднее (окно 2):\n", s_rolling)

# 4. Строим два графика: реальных продаж и сглаженного значения
plt.figure(figsize=(10, 5))
plt.plot(s_sales.index, s_sales.values, marker="o", linewidth=2, label="Реальные продажи")
plt.plot(s_rolling.index, s_rolling.values, marker="s", linestyle="--", linewidth=2, label="Скользящее среднее (rolling(2))")

plt.title("Продажи в интернет-магазине и скользящее среднее")
plt.xlabel("Дата")
plt.ylabel("Продажи")
plt.grid(True)
plt.xticks(rotation=30)
plt.legend()

plt.tight_layout()
plt.show()
```
